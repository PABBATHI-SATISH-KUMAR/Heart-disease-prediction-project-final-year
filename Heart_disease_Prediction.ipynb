# KNN

````python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
import joblib
import os

# Load or train model and scaler
model_path = 'cvd_knn_model.pkl'
scaler_path = 'cvd_knn_scaler.pkl'

if not os.path.exists(model_path) or not os.path.exists(scaler_path):
    # Load dataset and train model if not already saved
    df = pd.read_csv('heart_cleveland.csv')
    features = ['age', 'chol', 'sex', 'trestbps', 'cp', 'fbs', 'thalach']
    X = df[features]
    y = df['target']

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    model = KNeighborsClassifier(n_neighbors=5)
    model.fit(X_scaled, y)

    joblib.dump(model, model_path)
    joblib.dump(scaler, scaler_path)
else:
    model = joblib.load(model_path)
    scaler = joblib.load(scaler_path)

# Features in order
feature_names = ['age', 'chol', 'sex', 'trestbps', 'cp', 'fbs', 'thalach']
prompts = {
    'age': "Enter your age: ",
    'chol': "Enter your serum cholesterol level (mg/dl): ",
    'sex': "Enter your gender (1 = male, 0 = female): ",
    'trestbps': "Enter your resting blood pressure (mm Hg): ",
    'cp': "Enter chest pain type (0 = typical angina, 1 = atypical angina, 2 = non-anginal pain, 3 = asymptomatic): ",
    'fbs': "Is fasting blood sugar > 120 mg/dl? (1 = yes, 0 = no): ",
    'thalach': "Enter your maximum heart rate achieved: "
}

# Collect inputs
user_inputs = []
for feature in feature_names:
    value = float(input(prompts[feature]))
    user_inputs.append(value)

# Make prediction
input_df = pd.DataFrame([user_inputs], columns=feature_names)
input_scaled = scaler.transform(input_df)
prediction = model.predict(input_scaled)[0]

# Output result
if prediction == 1:
    print(" ✅Heart Disease Detected")
else:
    print(" ❌Heart Disease Not Detected")

````

#Naive Bayes

````python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
import joblib
import os

# File paths for saving/loading the model and scaler
model_path = 'cvd_nb_model.pkl'
scaler_path = 'cvd_nb_scaler.pkl'

# Check if model and scaler already exist
if not os.path.exists(model_path) or not os.path.exists(scaler_path):
    # Load dataset
    df = pd.read_csv('heart_cleveland.csv')

    # Features and target variable
    features = ['age', 'sex', 'chol', 'trestbps', 'cp', 'fbs', 'thalach']
    X = df[features]
    y = df['target']

    # Scale the features
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Train Naive Bayes model
    model = GaussianNB()
    model.fit(X_scaled, y)

    # Save the trained model and scaler
    joblib.dump(model, model_path)
    joblib.dump(scaler, scaler_path)
else:
    # Load the pre-trained model and scaler
    model = joblib.load(model_path)
    scaler = joblib.load(scaler_path)

# Define input prompts
feature_names = ['age', 'sex', 'chol', 'trestbps', 'cp', 'fbs', 'thalach']
prompts = {
    'age': "Enter your age: ",
    'sex': "Enter your gender (1 = male, 0 = female): ",
    'chol': "Enter your serum cholesterol level (mg/dl): ",
    'trestbps': "Enter your resting blood pressure (mm Hg): ",
    'cp': "Enter chest pain type (0 = typical angina, 1 = atypical angina, 2 = non-anginal pain, 3 = asymptomatic): ",
    'fbs': "Is fasting blood sugar > 120 mg/dl? (1 = yes, 0 = no): ",
    'thalach': "Enter your maximum heart rate achieved: "
}

# Collect user input
user_inputs = []
for feature in feature_names:
    value = float(input(prompts[feature]))
    user_inputs.append(value)

# Make prediction
input_df = pd.DataFrame([user_inputs], columns=feature_names)
input_scaled = scaler.transform(input_df)
prediction = model.predict(input_scaled)[0]

# Display result
if prediction == 1:
    print("✅ Heart Disease Detected")
else:
    print("❌Heart Disease Not Detected")
````
